## 3. 데이터 기초 통계량

- **중심의 위치측도**
    - `표본평균 (Sample Mean)`
        - 중심을 나타내는 측도 중에서 가장 많이 사용되는 방법
        - 자료의 무게 중심을 나타냄
        - $\overline x = \frac{\sum_{i = 1}^{n} x_i}{n}$
        - 자료의 이상치 (Outlier)에 영향을 많이 받음 (Trimmed mean)
    
    - `중앙값 (Median)`
        - 전체 관측값을 크기 순으로 나열한 했을 때 중앙에 위치한 값
        - 자료의 갯수(n) 이 홀수이면 $\frac{n+1}{2}$ 번째 관측값
        - 자료의 갯수(n) 이 짝수이면 $\frac{n}{2}$과 $\frac{n+1}{2}$ 번째 관측값의 평균값
        - 자료의 이상치 (Outlier)에 영향을 적게 받음

- **DataFrame 메소드**
    - `.mean()`: DataFrame/Series 의 평균, titanic["Fare"].mean()
    - `.median()`: DataFrame/Series 의 중앙값, titanic["Fare"].median()

```python
fare_mean = titanic["Fare"].mean()
print("fare_mean = ", fare_mean)
```

```python
fare_median = titanic["Fare"].median()
print("fare_median = ", fare_median)
```

- **분산과 표준편차**
    - 자료가 중심으로 부터  얼마나 퍼져있는 지를 표현 수치
    - $S^2 = \frac{\sum_i^N (X_{i}-\overline{X})^2}{N}$  
  <br />
    - sd = $\sqrt{S^2} = S$
    - 범위(Range): 최댓값 - 최솟값 

```python
## For 구문을 이용한 titanic 요금 분산

titanic_fare = titanic["Fare"]
# print("titanic_fare = ", titanic_fare)

fare_sum = 0
for i in titanic_fare:
    fare_sum = fare_sum + i

fare_mean = fare_sum/891

fare_var = 0

for i in titanic_fare:
    fare_var = fare_var + (i - fare_mean)**2

fare_var = fare_var/891

print("Variance of fare = ", fare_var)
```

```python
## For 구문을 이용한 titanic 요금 표준편차

fare_std = np.sqrt(fare_var)
print("std of fare = ", fare_std)
```

```python
## titanic 요금 분위 범위
fare_range = titanic_fare.max() - titanic_fare.min()
print("range of fare = ", fare_range)
```

```python
## DataFrame method를 이용한 분산, 표준편차, 범위

titanic_fare = titanic["Fare"]

titanic_fare_var = titanic_fare.var()
titanic_fare_std = titanic_fare.std()

print("titanic_fare.var() = ", titanic_fare_var)
print("titanic_fare.std() = ", titanic_fare_std)
```

* 결과 DataFrame으로 정리하기 

```python
summary_stat = {
    "mean": fare_mean,
    "median": fare_median,
    "variance": fare_var,
    "std": fare_std,
    "range": fare_range
}

display(summary_stat)
```

```python
df_summary = pd.DataFrame([summary_stat], index=['Fare'])
display("titanic 요금 통계량 = ", df_summary.round(3))
```

## 4. 데이터 정제 및 전처리
- **메서드란**: 파이썬의 함수작동 형식, 클래스내에 정의된 함수
- **정제 및 전처리 메서드**:
  - `.isnull()`: $\text{DataFrame}$의 각 요소가 **결측치**($\text{NaN}$)인지 여부를 불리언 $\text{DataFrame}$으로 반환
  - `.dropna()`: 결측치가 있는 행 또는 열을 제거
  - `.fillna()`: 결측치를 특정 값으로 채움
  - `.drop()`: 행 또는 열을 제거

## 5. 데이터 조작 및 계산
- 데이터를 변형하거나 통계적 계산을 수행하는 데 사용
- **조작 및 계산 메서드**:
  - `.sort_values()`: 하나 이상의 열을 기준으로 정렬
  - `.sort_indexs()`: 하나 이상의 인덱스를 기준으로 정렬
  - `.replace()`: DataFrame 또는 $\text{Series}$의 특정 값을 다른 값으로 변경
  - `.groupby()`: 특정 열을 기준으로 데이터를 그룹화 

- **< mtcars >** 자료 소개 
- mtcars는 통계 소프트웨어 R에 내장된 데이터셋으로, 데이터 분석을 위한 예제 자료로 널리 사용됩니다. 이 데이터셋은 1974년 미국 잡지 '모터 트렌드 US(Motor Trend US)'에 실린 내용을 기반으로 하며, 1973~1974년식 자동차 32대에 대한 연료 소비량 및 10가지 자동차 디자인 및 성능 측면의 정보를 포함하고 있습니다. 
- mtcars 데이터셋은 32개의 관측치(자동차 모델)와 11개의 변수로 구성되어 있습니다. 각 변수는 다음과 같은 정보를 담고 있습니다.

    - `mpg`: 마일/갤런(연비). 1갤런당 주행할 수 있는 마일 수
    - `cyl`: 실린더 수
    - `disp`: 배기량 (입방인치)
    - `hp`: 총 마력
    - `drat`: 후륜 차축 비율
    - `wt`: 중량 (1000파운드)
    - `qsec`: 1/4마일 도달 시간 (가속도 측정)
    - `vs`: 엔진 유형 (0 = V-형, 1 = 직렬형)
    - `am`: 변속기 유형 (0 = 자동, 1 = 수동)
    - `gear`: 전진 기어 수
    - `carb`: 기화기 수

```python
mtcars_path = os.path.join(data_dir, "mtcars.xlsx")
mtcars = pd.read_excel(mtcars_path, index_col=0)
mtcars.head()
```

### 열 값 기준으로 정렬 (.sort_values())

```python
# mpg (연비) 기준으로 내림차순 정렬 (ascending=False)
sorted_by_mpg = mtcars.sort_values(by='mpg', ascending=False) # by='컬럼명'
# sorted_by_mpg = mtcars.sort_values("mpg", ascending=False)

print("\n--- .sort_values() (mpg 내림차순 정렬) ---")
print(sorted_by_mpg.head())
```

```python
# cyl, mpg (연비) 기준으로 내림차순 정렬 (ascending=False)
sorted_by_mpg = mtcars.sort_values(by=['cyl','mpg'], ascending=False)

print("\n--- .sort_values() (['cyl','mpg'] 내림차순 정렬) ---")
print(sorted_by_mpg.head())
```

### 인덱스 기준으로 정렬 (.sort_index())

```python
# 인덱스 (차량 이름) 기준으로 오름차순 정렬 (기본값: ascending=True)
sorted_by_index = mtcars.sort_index()

print("\n--- .sort_index() (차량 이름 오름차순 정렬) ---")
print(sorted_by_index.head(10))
```

### <열값 정렬 문제>
- `mtcars` 자료를 무게 (wt)에 따라서 오름차순으로 정리하고 자료를 mtcars_wt로 저장한 후 10 ~ 20 열에 있는 자료만 프린트 하시오

```python
# < 풀이 >

mtcars_wt = mtcars.sort_values("wt", ascending=True)
mtcars_wt.head()
```

### 값 대체 (.replace())

```python
# gear 열의 숫자 값을 문자열로 대체
# 3 -> 'Low Gear', 4 -> 'Mid Gear', 5 -> 'High Gear', 6 -> 'Sport Gear'
mtcars['gear_label'] = mtcars['gear'].replace({
    3: 'Low Gear',
    4: 'Mid Gear',
    5: 'High Gear',
    6: 'Sport Gear'
})

print("\n--- .replace() (gear 값 대체) ---")
print(mtcars[['gear', 'gear_label']].head(7))
```

### <값 대체 문제>
- `mtcars` 자료에서 **am** 열의 값을 `0 -> 수동, 1 -> 자동` 으로 변경하고 프린트 하시오

```python
# < 풀이 >

mtcars["am_rename"]= mtcars["am"].replace({0:"자동", 1:"수동"})
print(mtcars.head())
```

### 데이터 그룹화 및 집계 (.groupby())

```python
# 'cyl' (실린더 개수) 기준으로 그룹화하고 'mpg'의 평균 계산
grouped_mpg = mtcars.groupby('cyl')['mpg'].mean()

print("\n--- .groupby() (실린더별 평균 mpg) ---")
print(grouped_mpg)
#grouped_mpg
```

```python
# 'cyl' (실린더 개수) 기준으로 그룹화하고 'mpg'의 중앙값 계산
grouped_mpg = mtcars.groupby('cyl')['mpg'].median()

print("\n--- .groupby() (실린더별 중앙값 mpg) ---")
print(grouped_mpg)
```

```python
# 'cyl' (실린더 개수) 기준으로 그룹화하고 'mpg'의 표준편차를 계산
grouped_mpg = mtcars.groupby('cyl')['mpg'].std()

print("\n--- .groupby() (실린더별 표준편차 mpg) ---")
print(grouped_mpg)
```

```python
# 'am', 'cyl' (실린더 개수) 기준으로 그룹화하고 'mpg'의 평균 계산
grouped_mpg = mtcars.groupby(['am', 'cyl'])['mpg'].mean()

print("\n--- .groupby() (실린더별 평균 mpg) ---")
print(grouped_mpg)
```

## 데이터 그룹화 및 집계 (언패킹 연산자 / 구조분해 할당)
- 언패킹 연산자(*) 와 구조분해 할당 이용

```python
# 언패킹 연산자
grouped_mpg = mtcars.groupby('cyl')
print(*grouped_mpg)

# 구조분해 할당
cyl_4, cyl_6, cyl_8 = grouped_mpg
print(cyl_4, cyl_6, cyl_8, sep="/n")
```

### <데이터 그룹화 및 집계 문제>
- `mtcars` 자료에서 **엔진유형 (vs)** 에 따른 **연비 (mpg)** 의 평균값과 표준편차를 각 각 구하고 프린트 하시오

```python
# < 풀이 >
print("mean by am = ")
print(mtcars.groupby("am")["mpg"].mean())

print("\n std by am = ")
print(mtcars.groupby("am")["mpg"].std())
```

## 6. 데이터 인덱싱
- **인덱싱할 열의 값은 고유한 것이 좋습니다.**
- **데이터 인덱싱 메서드**:
  - `.reset_index()`: 현재 인덱스를 열로 변환하고, 새로운 기본 정수 인덱스를 설정
  - `.set_index()`: $\text{DataFrame}$의 특정 열을 새로운 인덱스로 설정
  - `.rank()`: 데이터의 순위를 매김

### 인덱스를 열로 변환 (.reset_index())

```python
# 현재 인덱스(차량 이름)를 'CarName'이라는 열로 변환
mtcars_reset = mtcars.reset_index(names=['CarName'], drop = False)

print("\n--- .reset_index() 적용 결과 ---")
print(mtcars_reset.head())
```

### 특정 열을 인덱스로 설정 (.set_index())

```python
# 'CarName' 열을 제거하고 'cyl' 열을 새로운 인덱스로 설정
mtcars_set = mtcars_reset.set_index('cyl')

print("\n--- set_index() 적용 결과 (cyl 기준) ---")
print(mtcars_set)
print(f"\n새로운 인덱스 이름: {mtcars_set.index.name}")
```

### <인덱스 바꾸기 문제>
- 위 예제에서 구해던 `mtcars_set`을 이용하여 0 부터 시작하는 정수형 인덱스로 바꾸고 기존의 인덱스 cyl을 컬럼으로 복귀시키고 자료를 `mtcars_reset`으로 저장하고 출력해 보시오

```python
# < 풀이 >

mtcars_reset = mtcars_set.reset_index("cyl")
display(mtcars_reset.head())
```

### 데이터 순위 매기기 (.rank())

```python
# hp가 낮은 순서대로 순위를 매김 (오름차순). 동점은 평균 순위를 부여.
# 62hp가 가장 낮으므로 1위
mtcars['hp_rank_asc'] = mtcars['hp'].rank(ascending=True, method='average')

print("\n--- 3.1. .rank() (hp 오름차순 순위) ---")
print(mtcars[['hp', 'hp_rank_asc']].sort_values('hp', ascending=False).head())
```

```python
# hp가 높은 순서대로 순위를 매김 (내림차순). 동점은 낮은 순위(min)를 부여.
# 245hp가 가장 높으므로 1위
mtcars['hp_rank_desc'] = mtcars['hp'].rank(ascending=False, method='min')

print("\n--- .rank() (hp 내림차순 순위, method='min') ---")
print(mtcars[['hp', 'hp_rank_desc']].sort_values('hp', ascending=False).head())
```

### <데이터 순위 매기기 문제>
- `mtcars` 자료를 이용하여 **연비 (mpg)** 에 따라서 가장 높은 값이 1 갖도록 만들고 순위자료를 `mtcars` 에 **ranked** 라는 열이름으로 저장하고 출력해 보시오

```python
# < 풀이 >
mtcars["ranked"] = mtcars["mpg"].rank(ascending=False)
display(mtcars)
```
