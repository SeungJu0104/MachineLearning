### Multichain

```python
## Multichain
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. Chat 모델을 사용합니다.
# ---
chat_model = ChatOpenAI(model="gpt-4o-mini")

# 2. ChatPromptTemplate 정의
# ---
# 체인 1: 한국어 -> 영어 번역
prompt1 = ChatPromptTemplate.from_template(
    "다음에 나오는 단어 '{korean_word}'를 영어로 번역해 주세요. 번역된 단어만 출력하세요."
)

# 체인 2: 영어 -> 한국어 설명
prompt2 = ChatPromptTemplate.from_template(
    "다음의 영어 단어 '{english_word}'를 한국어로 30자 내외로 쉽게 설명해 주세요."
)

# 3. 파서 설정
# ---
# 문자열 결과만 추출하는 파서
parser = StrOutputParser()

# 4. LCEL 체인 구성
# ---

# 체인 1: 한국어 단어를 받아 영어 번역을 출력
# 입력: {"korean_word": "..."}
# 출력: "Insomnia"
chain1 = prompt1 | chat_model | parser

# 체인 2: 영어 단어를 받아 한국어 설명을 출력
# 입력: {"english_word": "..."}
# 출력: "잠을 쉽게 이루지 못하거나 충분히 자지 못하는 수면 장애입니다."
chain2 = prompt2 | chat_model | parser

# 4. 체인 실행 및 결과 출력
# ---

korean_word_input = "오류역전파"

# 체인 1 실행: 한국어 -> 영어 번역
print(f"--- 1. 한국어 '{korean_word_input}' -> 영어 번역 ---")
english_result = chain1.invoke({"korean_word": korean_word_input})
print(f"영어 번역 결과: {english_result}")
print("-" * 40)

# 체인 2 실행: 영어 -> 한국어 설명
print(f"--- 2. 영어 '{english_result}' -> 한국어 설명 ---")
korean_explanation = chain2.invoke({"english_word": english_result})
display(f"한국어 설명 결과: {korean_explanation}")
print("-" * 40)
```

```python
## Multichain 연결
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


# 1. 프롬프트 및 모델 정의 (이전 단계와 동일)
# ---

# LLM 설정
# gpt-4o-mini는 chat_model 변수로 유지합니다.
chat_model = ChatOpenAI(model="gpt-4o-mini")
parser = StrOutputParser()

# Prompt 1: 한국어 -> 영어 번역 (출력에 번역된 단어만 남도록 명시)
prompt1 = ChatPromptTemplate.from_template(
    "다음에 나오는 단어 '{korean_word}'를 영어로 번역해 주세요. 번역된 영어 단어만 출력하세요."
)

# Prompt 2: 영어 -> 한국어 설명
prompt2 = ChatPromptTemplate.from_template(
    "다음의 영어 단어 '{english_word}'를 한국어로 30자 내외로 쉽고 친절하게 설명해 주세요."
)

# 2. 체인 1 (번역 체인)
# ---

# chain1: 한국어 단어를 받아 영어 단어로 변환합니다.
chain1 = prompt1 | chat_model | parser

# 3. 체인 2 (통합 체인: 번역 결과 재활용)
# ---

# LCEL을 사용하여 chain1의 출력을 다음 단계의 입력 딕셔너리 키("english_word")에 매핑합니다.
#
# **동작 원리:**
# 1. chain2.invoke({"korean_word": "오류역전파"})가 실행됩니다.
# 2. {"english_word": chain1}이 먼저 실행됩니다. 이때 chain1은 "오류역전파"을 입력받아 "Backpropagation"를 출력합니다.
# 3. 결과 딕셔너리는 {"english_word": "Backpropagation"}가 됩니다.
# 4. 이 딕셔너리가 prompt2로 전달되어 {english_word} 변수가 채워집니다.
# chain1 과 chain2 를 연결해 실행(MultiChain)
chain2_integrated = ( 
    {"english_word": chain1} # {'english_word':'back propagation'} | prompt2 | chat_model | parser
    | prompt2
    | chat_model
    | parser
    # chain1 실행결과가 딕셔너리에 들어가고, 나머지 체인들이 실행
)

# 4. 체인 실행 및 결과 출력
# ---

korean_input = "오류역전파"
print(f"--- 통합 체인 실행: '{korean_input}' ---")

final_explanation = chain2_integrated.invoke({"korean_word": korean_input})

print(f"\n입력 단어: {korean_input}")
print(f"\n번역 단어: {chain1}")
print(f"최종 한국어 설명 결과: {final_explanation}")
```

### LLM parameters
  - `Temperature`: 생성된 텍스트의 다양성, 높을 수록 다양한 문장생성
  - `Max Tokens` (최대 토큰 수): 생성할 최대 토큰 수 (문장의 길이)
  - `Top P` (Top Probability): 다음 토큰을 생성할 때 상위 P% 토큰만을 고려
  - `Frequency Penalty` (빈도 패널티): 값이 클수록 이미 등장한 단어나 구절이 다시 등장할 확률을 감소 (0~1)
  - `Presence Penalty` (존재 패널티): 값이 클수록 아직 텍스트에 등장하지 않은 새로운 단어의 사용이 장려됩니다. (0~1)
  - `Stop Sequences` (정지 시퀀스): 특정 단어나 구절이 등장할 경우 생성을 멈추도록 설정. 리스트 형태 (예, terror, IS)

```python
## chat model with llm parameters

chat_model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7,         # 생성된 텍스트의 다양성 조정
    max_tokens=10,          # 생성할 최대 토큰 수
    frequency_penalty = 0.5,  # 이미 등장한 단어의 재등장 확률
    presence_penalty = 0.5,   # 새로운 단어의 도입을 장려
    stop = ["\n"]             # 정지 시퀀스 설정
)

# 모델 호출
question = "What is Alzhiemer's dementia?"
response = chat_model.invoke(question)

# 전체 응답 출력
print(response)
```

## Tool calling
- 대규모 언어 모델(LLM)이 외부 세계와 상호작용하고 자체 기능 범위를 확장할 수 있도록 도와주는 `외부 도구 또는 함수`
- LLM은 사용자의 요청을 이해하고, 이를 수행하기 위해 필요한 툴을 선택하고 호출하며, 툴의 실행 결과를 다시 받아 최종 답변을 생성하는 방식으로 작동
- 예를 들어, 날씨 정보 조회, 데이터베이스 조회, 계산기 사용, API 호출 등
- **도구 호출 방법**
  - 1. 도구 정의 (Tool Definition)
  - 2. 모델에 도구 바인딩 (Bind Tools to Model)
  - 3. 모델 호출 (Invoke the Model)
  - 4. 도구 실행 및 결과 반환 (Execute Tool and Get Result)

```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, ToolMessage


# 1. 단일 도구 정의 (Tool Definition)
@tool
def multiply(a: int, b: int) -> int:
    """
    두 정수 a와 b를 곱합니다. 이 도구는 복잡한 곱셈 계산에 사용됩니다.
    Args:
        a: 첫 번째 정수
        b: 두 번째 정수
    """ 
    # description 기술해야 작업 수행
    return a * b


tools = [multiply]
display("tools = ", tools)
```

```python
# 2. 모델에 도구 바인딩 (Bind Tool to Model)
# 도구 호출을 지원하는 모델을 사용해야 합니다. (예: gpt-4o, gemini-2.5-flash 등)
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
llm_with_tools = llm.bind_tools(tools)

user_query = "137에 45를 곱하면 얼마야?"

print("--- LLM 응답 (도구 호출 없음) ---")
llm_msg = llm.invoke(user_query)
display("llm msg = ", llm_msg)

# 3. LLM에게 도구 호출이 필요한 질문 던지기 (First Call: Request Tool Call)
user_query = "137에 45를 곱하면 얼마야?"
ai_msg = llm_with_tools.invoke(user_query)

print("--- LLM의 첫 번째 응답 (도구 호출 요청) ---")
display("ai_msg = ", ai_msg)
print(f"Tool Calls: {ai_msg.tool_calls}")
```

```python

# 4. 도구 실행 및 결과를 LLM에게 다시 전달 (Second Call: Execute and Pass Result)
if ai_msg.tool_calls:
    print("\n--- 도구 실행 ---")

    tool_call = ai_msg.tool_calls[0]
    tool_name = tool_call['name']
    tool_args = tool_call['args']
    tool_call_id = tool_call['id']
    tool_result = multiply.invoke(tool_args)

    print(f"실행할 도구: {tool_name}")
    print(f"입력 인수: {tool_args}")
    print(f"도구 ID: {tool_call_id}")
    print(f"결과: {tool_result}")

    # 도구 실행 결과를 ToolMessage 형태로 LLM에게 다시 전달합니다.
    tool_message = ToolMessage(
        content=str(tool_result), # 응답 결과
        tool_call_id=tool_call_id # tool_call ID
    )

    # 5. LLM에게 최종 답변 요청 (Final Call: Get Final Answer)
    # 이전 대화 내용(사용자 질문, LLM의 도구 요청, 도구 실행 결과)을 모두 전달합니다.
    messages_list = [
        HumanMessage(content=user_query),
        ai_msg,
        tool_message
    ]

    # display("messages_list = ", messages_list)
    final_response = llm_with_tools.invoke(messages_list)

    print("\n--- LLM의 최종 응답 (결과 정리) ---")
    print(final_response.content)
```

```python
## 2개 이상 tool calling

from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# 1. 도구 정의 (Tool Definition)
@tool
def multiply(a: int, b: int) -> int:
    """
    두 정수 a와 b를 곱합니다.
    Args:
        a: 첫 번째 정수
        b: 두 번째 정수
    """
    return a * b

@tool
def add(a: int, b: int) -> int:
    """
    두 정수 a와 b를 더합니다.
    Args:
        a: 첫 번째 정수
        b: 두 번째 정수
    """
    return a + b

tools = [multiply, add]

# 2. 모델에 도구 바인딩 (Bind Tools to Model)

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
llm_with_tools = llm.bind_tools(tools)

# 3. 모델 호출 (Invoke the Model)
# LLM은 사용자의 요청을 분석하여 어떤 도구를 어떤 인수로 호출할지 결정합니다.
query = "123에 456을 곱한 값에 789를 더한 값은 얼마야?"
llm_msg = llm.invoke(query)

print("--- LLM의 응답 (도구 호출 없음) ---")
print(llm_msg.content)

ai_msg = llm_with_tools.invoke(query)

print("--- LLM의 응답 (도구 호출 요청) ---")
print(ai_msg.tool_calls) # LLM이 호출하기로 결정한 도구와 인수 확인

# 4. 도구 실행 및 결과 반환 (Execute Tool and Get Result)
# 실제 애플리케이션에서는 이 단계에서 도구를 실행하고 결과를 LLM에 다시 전달합니다.

# 예제 실행
if ai_msg.tool_calls:
    print("\n--- 도구 실행 결과 ---")

    # 첫 번째 도구 호출 실행 (123 * 456)
    call_1 = ai_msg.tool_calls[0]
    result_1 = multiply.invoke(call_1['args'])
    print(f"도구 '{call_1['name']}' 실행: 결과 {result_1}")

    # 두 번째 도구 호출 실행 (결과 + 789를 의도했으나, 예제 모델은 789만 더하는 것으로 가정)
    # 실제 복잡한 계산은 에이전트(Agent) 워크플로우를 통해 체이닝되어야 합니다.
    call_2 = ai_msg.tool_calls[1] # 워크플로우 체이닝

    call_2['args']['b'] = result_1
    result_2 = add.invoke(call_2['args'])
    print(f"도구 '{call_2['name']}' 실행: 결과 {result_2}")
```

```python
from langchain_community.tools import TavilySearchResults
from langchain_core.tools import tool

# Tool 정의
@tool
def search_web(query: str) -> str:
    """Searches the internet for information that does not exist
    in the database or for the latest information.
    """

    tavily_search = TavilySearchResults(max_results=2)
    docs = tavily_search.invoke(query) # 검색 결과

    formatted_docs = "\n---\n".join([
        f'<Document href="{doc["url"]}"/>\n{doc["content"]}\n</Document>'
        for doc in docs
        ])

    if len(formatted_docs) > 0:
        return formatted_docs

    return "관련 정보를 찾을 수 없습니다."
```

## Tool calling 결과를 이용한 LLM 답변 생성

```python
from datetime import datetime
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableConfig, chain
from langchain_core.tools import tool

# 오늘 날짜 설정
today = datetime.today().strftime("%Y-%m-%d")

# 프롬프트 템플릿
prompt = ChatPromptTemplate.from_messages([
    ("system", f"You are a helpful AI assistant. Today's date is {today}."),
    ("placeholder", "{messages}"),
    ("human", "{user_input}"),
])

# ChatOpenAI 모델 초기화
llm = ChatOpenAI(model="gpt-4o-mini", temperature = 0)

# LLM에 도구를 바인딩
llm_with_tools = llm.bind_tools(tools=[web_search])

# LLM 체인 생성
llm_chain = prompt | llm_with_tools

# 도구 실행 체인 정의
@tool
def web_search_chain(user_input: str):
    '''
    Tool decorator demo
    '''
    inputs = {"user_input": user_input}
    ai_msg = llm_chain.invoke(inputs)
    display("ai_msg: \n", ai_msg)
    print("-"*100)
    tool_msgs = web_search.batch(ai_msg.tool_calls)
    display("tool_msgs: ", tool_msgs)
    print("-"*100)
    return llm_chain.invoke({**inputs, "messages": [ai_msg, *tool_msgs]})

# 체인 실행
response = web_search_chain.invoke("오늘 금의 가격은 얼마인가요?")

```

```python
# 응답 출력
print(f'{today}\n, {response.content}')
```
